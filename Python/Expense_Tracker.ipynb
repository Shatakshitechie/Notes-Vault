4 — Expense Tracker

```python
import logging
from functools import reduce
```

```python
logging.basicConfig(filename="expense_tracker.log", level=logging.INFO, force=True)
```

```python
expenses = []
```

Load the saved expenses from the file (if any)

```python
def load_expenses():
  try:
    with open("expenses.txt", 'r') as file:
      for line in file:
        name, amount = line.strip().split(',')
        expenses.append([name, float(amount)])
    logging.info(f"{len(expenses)} expenses loaded from file")
  except FileNotFoundError:
    logging.info("No saved file found, starting fresh")
```

```python
def add_expense(name, amount):
  expenses.append([name, amount])
  print("Expense added successfully...!")
  logging.info(f"Expense added: {name} - {amount}")
```

```python
def view_all():
  if len(expenses) == 0:
    print("No expenses found.")
  else:
    for i, e in enumerate(expenses, 1):
      print(f"{i}. {e[0]} - Rs.{e[1]}")
  logging.info("Viewed all expenses")
```

```python
def total_expense():
  amounts = [e[1] for e in expenses]
  if len(amounts) == 0:
    return 0
  return reduce(lambda a,b: a+b, amounts)
```

```python
def above_1000():
  return list(filter(lambda a: a[1]>1000, expenses))
```

```python
def highest_expense():
  return reduce(lambda a,b: a if a[1]>b[1] else b, expenses)
```

```python
def save_expenses():
  with open("expenses.txt", 'w') as file:
    for e in expenses:
      file.write(f"{e[0]},{e[1]}\n")
  print("Expenses saved to expenses.txt")
  logging.info(f"{len(expenses)} expenses saved to file")
```

```python
logging.info("Program started")
load_expenses()

while True:
  try:
    choice = int(input("\nChoose: 1.Add  2.View  3.Total  4.Above 1000\n5.Highest  6.Save  7.Exit => "))

    if choice == 1:
      logging.info("User wants to add an expense")
      name = input("Enter expense name: ").strip().title()
      if name == "" or "," in name:
        print("Name can't be empty or contain a comma!")
        logging.error(f"Invalid expense name entered: {name}")
        continue
      amount = float(input("Enter the amount: "))
      if amount <= 0:
        print("Amount must be greater than 0!")
        logging.error(f"Invalid amount entered: {amount}")
        continue
      add_expense(name, amount)

    elif choice == 2:
      view_all()

    elif choice == 3:
      total = total_expense()
      print(f"Total expenses = Rs.{total}")
      logging.info(f"Total expense calculated: {total}")

    elif choice == 4:
      big = above_1000()
      if len(big) == 0:
        print("No expense above Rs.1000.")
      else:
        for e in big:
          print(f"{e[0]} - Rs.{e[1]}")
      logging.info(f"Expenses above 1000 found: {len(big)}")

    elif choice == 5:
      if len(expenses) == 0:
        print("No expenses found.")
      else:
        high = highest_expense()
        print(f"Highest expense: {high[0]} - Rs.{high[1]}")
        logging.info(f"Highest expense found: {high[0]} - {high[1]}")

    elif choice == 6:
      save_expenses()

    elif choice == 7:
      save_expenses()
      print("Thank you for using the expense tracker!")
      logging.info("Program ended")
      break

    else:
      print("Wrong value entered....")
      logging.error(f"Invalid menu option: {choice}")

  except ValueError:
    print("Please enter a number only!")
    logging.error("Error: Value error")
```
