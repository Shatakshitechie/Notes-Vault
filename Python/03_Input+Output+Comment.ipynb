# Output (print Function):

Python uses the built-in `print()` function to display output on the screen.

- <small>**Syntax**: `print(value)` – displays whatever is passed inside the parentheses</small>
- <small>**Text/strings** must be wrapped in quotes (`" "` or `' '`)</small>

```python
print("Hello! It's Shatakshi.")
```

Numbers and variables can be passed directly

```python
print(6)
```

```python
a = "Hello"
print(a)
```

Multiple values can be printed together, separated by commas

```python
print("Age:", 24)
```

<span style="font-size:1.15em">**Printing Multiple Values & f-strings**</span>
There are a few ways to combine text and variables in `print()`:

```python
name = "Shatakshi"
age = 24
```

- <small>**Comma-separated**: Python adds a space between each value automatically</small>

```python
print("Name:", name, "Age:", age)
```

- <small>**Concatenation with `+`**: works only with strings; numbers must be converted first with `str()`</small>

```python
print("Name: " + name + ", Age: " + str(age))
```

- <small>**`.format()` method**: placeholders `{}` get filled in order by the arguments passed to `.format()`</small>

```python
print("Name: {}, Age: {}".format(name, age))
```

- <small>**f-strings** (formatted string literals): variables go directly inside `{}`, no `+`, `str()`, or `.format()` needed; introduced in Python 3.6 and now the preferred way to format strings</small>

```python
print(f"Name: {name}, Age: {age}")
```

# Input (input Function):

**1. Basic `input()`**
Whatever the user types is read and returned as a **string**, regardless of what it looks like — even if the user enters a number, Python treats it as text. An optional message can be passed inside `input()` to prompt the user before they type, making the program more user-friendly.

```python
name = input("Enter your name: ")
print(f"Hello, {name}!")
```

**2. Taking int/float input**
Since `input()` always returns a string, it can't directly be used in calculations — attempting math on it would throw an error. To take numeric input, the returned string must be explicitly converted using `int()` for whole numbers or `float()` for decimal values, wrapping the `input()` call itself.

```python
age = int(input("Enter your age: "))
height = float(input("Enter your height (in m): "))
print(f"Age: {age}, Height: {height}")
```

**3. Taking multiple values (including lists and tuples)**
Often a single line of input contains several values separated by spaces (or another character, like a comma). The `.split()` method breaks that input string into a list of substrings at each separator, which can then be:


- Unpacked directly into separate variables, when the number of values is known in advance

```python
a, b = input("Enter two numbers separated by space: ").split()
print(a, b)
```

- Converted all at once into integers or floats using `map()`, instead of converting each value individually

```python
x, y = map(int, input("Enter two integers separated by space: ").split())
print(x, y)
```

- Collected into a `list` or `tuple` using the respective constructor, when the number of values may vary or needs to be stored together

```python
num_list = list(map(int, input("Enter numbers separated by space: ").split()))
print(num_list)
```

- Extended to reading a fixed number of values across **multiple lines** as well, typically using a loop or list comprehension that repeats `input()` calls

```python
n = int(input("How many numbers? "))
values = [int(input()) for _ in range(n)]
print(values)
```

**NOTE**: some uses `eval(input())` as a shortcut to directly parse typed-out lists or tuples (e.g. typing `[1, 2, 3]`). This should be avoided — `eval()` executes whatever string it's given as code, which is a serious security risk if the input isn't fully trusted.

# Comment:

<span style="font-size:1.15em">**Comments in Python**</span>
Comments are ignored by Python when the code runs — they're notes for humans reading the code.

- <small>**Single-line**: starts with `#` – everything after it on that line is ignored

```python
# This is a single-line comment
print("Hello")  # can also follow code on the same line
```

- <small>**Multi-line**: Python has no true multi-line comment syntax; instead, a triple-quoted string (`'''...'''` or `"""..."""`) not assigned to anything is commonly used the same way, since it's written but never used</small>

```python
'''
This is a multi-line comment
spanning several lines
'''
print("Hello")
```
