# String Manipulation



<span style="font-size:1.15em">**Escape Characters**</span>
An escape character lets you include a special or otherwise hard-to-type character inside a string, by prefixing it with a backslash (`\`). Python interprets the backslash and the character right after it together as a single special instruction, rather than as two separate characters.

They are especially useful when a string needs to include a quote character matching its own delimiter, or when formatting output across multiple lines/columns without manually pressing Enter or Tab in the source code.

- <small>**`\n`** – newline, moves the following text to a new line</small>

```python
print("Line one\nLine two")
```

- <small>**`\t`** – tab, inserts a horizontal tab space</small>

```python
print("Name:\tAge:")
```

- <small>**`\\`** – a literal backslash, since a single `\` on its own would try to start an escape sequence</small>

```python
print("This is a backslash: \\")
```

- <small>**`\'`** – a literal single quote, useful when the string itself is wrapped in single quotes</small>

```python
print('It\'s a sunny day')
```

- <small>**`\"`** – a literal double quote, useful when the string itself is wrapped in double quotes</small>

```python
print("She said \"hello\"")
```

- <small>**`\b`** – backspace, removes the character immediately before it</small>

```python
print("Hello_\bWorld")
```

<span style="font-size:1.15em">**Indexing**</span>
Since a string is a sequence of characters, each character has a fixed position, called its **index**, that can be used to access it directly.

```python
text = "Hello World"
```

- <small>Indexing starts at **0**, not 1 – the first character is at index `0`, the second at index `1`, and so on</small>

```python
text[0]    # H  (first character)
```

```python
text[6]    # W  (7th character)
```

- <small>**Negative indexing** counts from the end of the string, starting at `-1` for the last character, `-2` for the second-last, and so on – useful when you want to access something from the end without knowing the string's exact length</small>

```python
text[-1]   # d  (last character)
```

```python
text[-5]   # W  (5th from the end)
```

- <small>Since a space is also a character, it occupies an index position just like any letter or digit</small>

```python
text[5]    # ' ' (the space is a valid, indexable character)
```

<small>**Note:** Accessing an index outside the string's range (in either direction) raises an `IndexError`</small>

<span style="font-size:1.15em">**Slicing**</span>
While indexing gives you a single character, slicing lets you extract a **range** of characters at once, producing a new substring.

- <small>**Syntax**: `string[start:stop]` – returns characters from index `start` up to, but **not including**, index `stop`</small>

```python
text[2:8]
```

- <small>Omitting `start` defaults to the beginning of the string; omitting `stop` defaults to the end</small>

```python
text[:8]
```

```python
text[3:]
```

- <small>A third value can be added — `string[start:stop:step]` – to skip characters, e.g. a step of `2` takes every second character</small>

```python
text[2:10:2]
```

- <small>Negative indices work in slicing too, allowing you to slice relative to the end of the string</small>

```python
text[10:2:-2]
```

- <small>A negative **step** reverses direction, which is also the common trick used to reverse an entire string</small>


```python
text[::-1]
```

- <small>Slicing never raises an error for out-of-range indices — it just returns as much as is available (unlike indexing, which raises `IndexError`)</small>

```python
text[3:12]
```

<span style="font-size:1.15em">**Concatenation**</span>
Concatenation means joining two or more strings together to form a single new string.

- <small>**`+` operator**: joins strings directly, e.g. combining a first and last name into a full name</small>
- <small>Only works between strings — attempting to concatenate a string with a number directly raises a `TypeError`; the number must first be converted using `str()`</small>
- <small>Since strings are immutable, concatenation doesn't modify either original string — it creates and returns a **new** string</small>
- <small>Spaces or punctuation aren't added automatically between concatenated strings — they must be included explicitly, e.g. `"Hello" + " " + "World"`</small>

```python
name = "Shatakshi"
surname = "Agarwal"
```

```python
name+surname
```

<span style="font-size:1.15em">**Repetition**</span>
The `*` operator repeats a string a given number of times, producing a new, longer string.

- <small>**Syntax**: `string * n` – repeats the string `n` times, where `n` is an integer</small>
- <small>Like concatenation, this creates a **new** string rather than modifying the original, since strings are immutable</small>

```python
"Hello!"*3
```

- <small>Multiplying by `0` (or a negative number) results in an empty string, not an error</small>

```python
"Hello!"*0
```

- <small>Often combined with concatenation for quick formatting, e.g. building a separator line</small>

```python
"-"*50
```

<span style="font-size:1.15em">**String Methods**</span>
Strings come with many built-in **methods** — functions called directly on a string using dot notation — for common transformations and checks. Since strings are immutable, every method below returns a **new** string (or value) rather than changing the original.

**A. Case Conversion**

```python
text = "hello World"
```

- <small>**`upper()` / `lower()`**: convert to all uppercase / all lowercase</small>

```python
text.upper()
```

```python
text.lower()
```

- <small>**`capitalize()`**: capitalizes only the first character, lowercases the rest</small>

```python
text.capitalize()
```

- <small>**`title()`**: capitalizes the first letter of every word</small>

```python
text.title()
```

- <small>**`swapcase()`**: swaps uppercase to lowercase and vice versa</small>

```python
text.swapcase()
```

- <small>**`casefold()`**: a more aggressive version of `lower()`, meant for case-insensitive comparison across languages; rarely needed outside that specific case</small>

```python
text.casefold()
```

**B. Whitespace & Alignment**

```python
padded = "   Hello   "
```

- <small>**`strip()` / `lstrip()` / `rstrip()`**: remove whitespace from both ends / left only / right only</small>

```python
padded.strip()
```

```python
padded.lstrip()
```

```python
padded.rstrip()
```

- <small>**`center(width)` / `ljust(width)` / `rjust(width)`**: pad the string to a given width, aligning center/left/right</small>

```python
"Hello".center(10, "-")
```

```python
"Hello".ljust(10, "-")
```

```python
"Hello".rjust(10, "-")
```

- <small>**`zfill(width)`**: pads a numeric string with leading zeros to reach a given width</small>

```python
"42".zfill(5)
```

**C. Searching**

```python
sentence = "I am learning Python"
```

- <small>**`find()` / `rfind()`**: index of the first/last occurrence of a substring; returns `-1` if not found</small>

```python
sentence.find("a")
```

```python
sentence.find("java")
```

```python
sentence.rfind("a")
```

- <small>**`index()` / `rindex()`**: same as `find()`/`rfind()`, but raise a `ValueError` instead of returning `-1`</small>

```python
sentence.index("a")
```

```python
sentence.index("java")
```

```python
sentence.rindex("a")
```

- <small>**`count()`**: number of times a substring appears</small>

```python
sentence.count("n")
```

- <small>**`startswith()` / `endswith()`**: check whether the string begins or ends with a given substring</small>

```python
sentence.endswith("Python")
```

```python
sentence.startswith("I am")
```

```python
sentence.endswith("Java")
```

**D. Splitting & Joining**

- <small>**`split()` / `rsplit()`**: split into a list of substrings, working from the left/right</small>

```python
fruits = "Apple,Banana,Oranges"
```

```python
fruits.split(",")
```

```python
fruits.split(",", 1)
```

```python
fruits.rsplit(",", 1)
```

- <small>**`splitlines()`**: splits at line breaks (`\n`)</small>

```python
fruits = "Apple\nBanana\nOranges"
```

```python
fruits.splitlines()
```

- <small>**`join()`**: joins a list of strings into one, using this string as the separator</small>

```python
fruits = ['Apple', 'Banana', 'Oranges']
```

```python
','.join(fruits)
```

- <small>**`partition()` / `rpartition()`**: splits into exactly 3 parts — before, the separator itself, and after — at the first/last occurrence</small>

```python
fruits = "Apple,Banana,Oranges"
```

```python
fruits.partition(',')
```

```python
fruits.rpartition(',')
```

**E. Replacing**
- <small>**`replace(old, new)`**: replaces all occurrences of a substring with another</small>


```python
sentence = "I am learning Java"
```

```python
sentence.replace('Java','Python')
```

**F. Content Checks** (commonly used for validating input)
- <small>**`isalpha()`**: only letters</small>

```python
"Hello123".isalpha()
```

```python
"Welcome".isalpha()
```

- <small>**`isdigit()` / `isdecimal()` / `isnumeric()`**: only numeric characters, with subtle differences in what counts as "numeric" (e.g. fractions, superscripts)</small>

```python
"123".isdigit()
```

```python
"Ⅷ".isdecimal()
```

```python
"Ⅷ".isnumeric()
```

- <small>**`isalnum()`**: only letters and/or digits</small>

```python
"Hello123".isalnum()
```

```python
"Hello@123".isalnum()
```

- <small>**`isspace()`**: only whitespace characters</small>

```python
"        ".isspace()
```

```python
"    a    ".isspace()
```

- <small>**`islower()` / `isupper()` / `istitle()`**: checks the string's current case</small>

```python
text = "Hello World"
```

```python
text.islower()    # True when all characters are lower case
```

```python
text.isupper()    # True when all characters are upper case
```

```python
text.istitle()   # True when all words start with an upper case character
```

- <small>**`isidentifier()`**: checks whether the string is valid as a Python variable name</small>

```python
"var_1".isidentifier
```

```python
"@_var1".isidentifier()
```

- <small>**`isascii()`**: only ASCII characters</small>

```python
print("café".isascii())
```

```python
print("cafe".isascii())
```

- <small>**`isprintable()`**: no non-printable/control characters</small>

```python
"Hello\tWorld".isprintable()
```

```python
"Hello World".isprintable()
```

**G. Advanced (optional)**
- <small>**`translate()` + `maketrans()`**: replace multiple characters at once using a translation mapping — useful for bulk operations like stripping all punctuation, but more advanced than the methods above</small>

```python
table = str.maketrans("abc", "xyz")
print("abcdef".translate(table))   # xyzdef
```

**H. Encoding**

```python
text = "Café"
```

- <small>**`encode()`**: converts a string into a `bytes` object, using a specified encoding (default is UTF-8) — needed when writing text to files/network in binary mode</small>

```python
encoded = text.encode()
```

```python
encoded
```

- <small>**`decode()`**: the reverse — converts a `bytes` object back into a string; technically a method on `bytes`, not `str`, but included here since it's the natural counterpart to `encode()`</small>

```python
encoded.decode()
```
