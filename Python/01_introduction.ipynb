# Introduction

<span style="font-size:1.15em">**1. What is Python?**</span>
Python is a high-level, general-purpose programming language known for its clean, readable syntax that lets code read almost like plain English.

It was designed by Guido van Rossum, who began working on it in the late 1980s as a successor to the ABC language, and released it publicly in 1991.

Since then, it has grown through community-driven development into one of the most widely used languages in the world, guided by a design philosophy - summed up in "The Zen of Python" - that favors simplicity and clarity over complexity. It has evolved through two major versions - Python 2 and Python 3 - with Python 3 being the current, actively maintained release.

<span style="font-size:1.15em">**2. What can Python do?**</span>
- <small>**Web development** – build web apps and APIs on the server side</small>
- <small>**Automation** – script workflows and repetitive tasks</small>
- <small>**Data handling** – connect to databases, read/write files</small>
- <small>**Data science & math** – process big data, run complex computations</small>
- <small>**Prototyping to production** – equally suited for quick prototypes and full-scale software</small>

<span style="font-size:1.15em">**3. Why learn Python?**</span>
- <small>**Cross-platform** – runs on Windows, Mac, Linux, Raspberry Pi and more, with little to no changes</small>
- <small>**Simple syntax** – close to plain English, easy to read and learn</small>
- <small>**Concise** – fewer lines of code needed compared to languages like Java</small>
- <small>**Interpreted** – code runs as soon as it's written, ideal for fast prototyping</small>
- <small>**Multi-paradigm** – supports procedural, object-oriented and functional programming</small>
- <small>**Rich ecosystem** – frameworks like Django and Flask for web development; Pandas, TensorFlow and Scikit-learn for data science and ML</small>
- <small>**Industry adoption** – trusted by companies like Google, Netflix and NASA</small>
- <small>**Career demand** – strong job opportunities in software development, data science and AI/ML</small>

<span style="font-size:1.15em">**4. How Python works**</span>

When you run a Python program, your code goes through a few stages before you see the result.

```mermaid
flowchart LR
    A["Source Code<br/>(.py)"] --> B["Compiler"]
    B --> C["Bytecode<br/>(.pyc)"]
    C --> D["Python Virtual Machine<br/>(Interpreter)"]
    D --> E["Output / Result"]
```

<span style="font-size:1.15em">**Why Python uses both a compiler and an interpreter**</span>
- <small>**Pure compiled languages** (like C) are fast, but produce machine code tied to a specific OS/platform, and need a separate build step before running</small>
- <small>**Pure interpreted languages** are portable and flexible, but slower, since they re-read and translate raw source code every single time they run</small>
- <small>**Python takes the middle path** – compiling once to platform-independent bytecode avoids re-parsing the raw source on every run, while interpreting that bytecode (instead of compiling all the way to machine code) keeps Python portable across platforms and preserves its dynamic, flexible nature</small>

This combination is what lets Python stay easy to write and run immediately, without giving up too much performance or portability.

<span style="font-size:1.15em">**5. Installation / Setup**</span>

- <small>**Step 1 – Download**: Go to the official site (python.org/downloads) and get the installer for your OS (Windows/Mac); most Linux distros come with Python pre-installed</small>
- <small>**Step 2 – Install**: Run the installer; on Windows, make sure to check **"Add Python to PATH"** before clicking install</small>
- <small>**Step 3 – Verify**: Open a terminal/command prompt and run `python --version` (or `python3 --version` on Mac/Linux) to confirm it's installed</small>
- <small>**Step 4 – Check pip**: Python comes bundled with **pip**, its package manager; verify with `pip --version`, used later to install libraries</small>
- <small>**Step 5 – Pick an editor/IDE**: Use the built-in **IDLE**, or a code editor like **VS Code**, **PyCharm**, or **Jupyter Notebook** for writing and running code</small>
- <small>**Step 6 – Run your first file**: Save a file as `hello.py` with `print("Hello, World!")`, then run it with `python hello.py`</small>
