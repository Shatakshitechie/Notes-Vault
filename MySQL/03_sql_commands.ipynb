# SQL Command Categories

SQL commands are instructions used to interact with a database — to define its structure, manipulate data, control access, or manage transactions. They're grouped into 5 categories:

- <small>**DDL** – Data Definition Language</small>
- <small>**DML** – Data Manipulation Language</small>
- <small>**DQL** – Data Query Language</small>
- <small>**DCL** – Data Control Language</small>
- <small>**TCL** – Transaction Control Language</small>

**1. DDL (Data Definition Language)**
Used to define or modify the structure of database objects (tables, schemas).
- <small>`CREATE` – creates a database/table</small>
- <small>`ALTER` – modifies an existing table structure</small>
- <small>`DROP` – deletes a database/table permanently</small>
- <small>`TRUNCATE` – removes all rows from a table, keeps structure</small>

**2. DML (Data Manipulation Language)**
Used to manipulate the actual data stored in tables.
- <small>`INSERT` – adds new data</small>
- <small>`UPDATE` – modifies existing data</small>
- <small>`DELETE` – removes data</small>

**3. DQL (Data Query Language)**
Used to retrieve/query data from the database.
- <small>`SELECT` – fetches data</small>

**4. DCL (Data Control Language)**
Used to control access and permissions on the database.
- <small>`GRANT` – gives user access privileges</small>
- <small>`REVOKE` – removes user access privileges</small>

**5. TCL (Transaction Control Language)**
Used to manage transactions (a set of operations treated as one unit).
- <small>`COMMIT` – saves all changes made in a transaction</small>
- <small>`ROLLBACK` – undoes changes made in a transaction</small>
- <small>`SAVEPOINT` – sets a point within a transaction to roll back to</small>

**Note:** DQL is sometimes grouped under DML, so you may see SQL commands classified into just 4 categories instead of 5.
