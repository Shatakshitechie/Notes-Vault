# Data Types

<span style="font-size:1.15em">**1. What are data types & why do we need them?**</span>
Data types define what kind of value a column can store (number, text, date, etc.), ensuring data is stored correctly, validated, and processed efficiently.

**2. What are the main categories of data types?**

- <small>**Numeric** – whole numbers, decimals (e.g. `INT`, `FLOAT`)</small>
- <small>**String/Text** – characters and text (e.g. `VARCHAR`, `TEXT`)</small>
- <small>**Date & Time** – dates, timestamps (e.g. `DATE`, `DATETIME`)</small>
- <small>**Boolean** – true/false values</small>
- <small>**Binary** – raw binary data like images/files (e.g. `BLOB`)</small>

**3. Numeric Types**

- <small>**INT** – whole numbers (e.g. 1, 100, -5)</small>
- <small>**FLOAT** – approximate decimal numbers, less precise, faster</small>
- <small>**DECIMAL** – exact decimal numbers, used for precision (e.g. money/currency)</small>

**4. String Types**

- <small>**CHAR** – fixed-length string (e.g. `CHAR(10)` always uses 10 characters, padded with spaces)</small>
- <small>**VARCHAR** – variable-length string, uses only as much space as needed (e.g. `VARCHAR(255)`)</small>
- <small>**TEXT** – for large blocks of text, no fixed length limit like VARCHAR</small>

**5. Date & Time Types**

- <small>**DATE** – stores date only (YYYY-MM-DD)</small>
- <small>**DATETIME** – stores date and time (YYYY-MM-DD HH:MM:SS)</small>
- <small>**TIMESTAMP** – like DATETIME, but auto-updates and is timezone-aware</small>
- <small>**TIME** – stores time only (HH:MM:SS)</small>

**6. Boolean Type**
MySQL doesn't have a true `BOOLEAN` type — it's stored as `TINYINT(1)`, where `0` = false and `1` = true. `BOOLEAN` is just an alias for `TINYINT(1)`.

**7. Binary Type (BLOB)**
Stores binary data such as images, audio, video, or files directly in the database.
- <small>**TINYBLOB** – up to 255 bytes</small>
- <small>**BLOB** – up to 65 KB</small>
- <small>**MEDIUMBLOB** – up to 16 MB</small>
- <small>**LONGBLOB** – up to 4 GB</small>
