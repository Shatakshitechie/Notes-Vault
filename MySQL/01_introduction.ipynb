# Introduction

**1. What is a data?**
A collection of raw, unprocessed facts, numbers, symbols, or observations.
It has no context or meaning, by organizing and analyzing data, it turns into useful information.

**1.1. Types of Data**
- <small>**Structured** – organized in rows/columns, like a spreadsheet</small>
- <small>**Semi-structured** – partially organized, like JSON or XML files</small>
- <small>**Unstructured** – freeform, like photos, videos, audio</small>
- <small>**Quantitative** – measurable, numeric</small>
- <small>**Qualitative** – descriptive, non-numeric</small>

**1.2. Data Preprocessing**
- <small>**Data Cleaning** – handling missing values, duplicates, errors</small>
- <small>**Data Transformation** – normalization, formatting, type conversion</small>
- <small>**Data Integration** – combining data from multiple sources</small>

**Why it matters?** Raw data is rarely usable as-is before storage/analysis.
