# Introduction

<span style="font-size:1.15em">**1. What is a data?**</span>
A collection of raw, unprocessed facts, numbers, symbols, or observations.
It has no context or meaning, by organizing and analyzing data, it turns into useful information.

**1.1. Types of Data**
- <small>**Structured** – organized in rows/columns, like a spreadsheet</small>
- <small>**Semi-structured** – partially organized, like JSON or XML files</small>
- <small>**Unstructured** – freeform, like photos, videos, audio</small>
- <small>**Quantitative** – measurable, numeric</small>
- <small>**Qualitative** – descriptive, non-numeric</small>

**1.2. Data Preprocessing**
- <small>**Data Cleaning** – handling missing values, duplicates, errors</small>
- <small>**Data Transformation** – normalization, formatting, type conversion</small>
- <small>**Data Integration** – combining data from multiple sources</small>

**Why it matters?** Raw data is rarely usable as-is before storage/analysis.

<span style="font-size:1.15em">**2. What is DataBase?**</span> An organized collection of digital data or information stored electronically in a computer system.

**2.1. Why do we need database?**

For proper storage and management of data, making it easy to organize, retrieve, and use for analysis.

**2.2. What are the types of databases?**

- <small>**Relational (SQL)** – structured data stored in tables, with defined relationships between them (e.g. MySQL, PostgreSQL, Oracle)</small>
- <small>**Non-relational (NoSQL)** – flexible, schema-less storage without fixed relationships (e.g. MongoDB, Firebase, Redis)</small>

<span style="font-size:1.15em">**3.Database Management?**</span> The process of organizing, storing, updating, retrieving, and securing data efficiently. Key tasks involved:
- <small>**Data Storage** – saving data in an organized structure</small>
- <small>**Data Retrieval** – fetching data when needed (queries)</small>
- <small>**Data Update** – modifying existing data</small>
- <small>**Data Security** – controlling access, preventing unauthorized use</small>
- <small>**Backup & Recovery** – protecting against data loss</small>

<span style="font-size:1.15em">**4. Database Management System(DBMS)?**</span> The software for performing management of database.
<small>**Eg.:** SQL, MySQL, Oracle, PostgreSQL, MongoDB, SQL Server</small>

**4.1. What are the types of DBMS?**

- <small>**Relational** – data stored in tables with relationships (e.g. MySQL)</small>
- <small>**Hierarchical** – data stored in a tree-like, parent-child structure</small>
- <small>**Network** – like hierarchical, but a record can have multiple parents (more flexible links)</small>
- <small>**Object-oriented** – data stored as objects, like in OOP</small>
- <small>**NoSQL** – flexible, schema-less storage for unstructured/semi-structured data</small>

**4.2. What is RDBMS?**
A type of DBMS that stores data in structured tables with relationships between them, using keys.
<small>**Eg.:** MySQL, PostgreSQL, Oracle, SQL Server</small>

**4.3. DBMS vs RDBMS**

- <small>**DBMS** – manages data, but doesn't require table-based structure or relationships (e.g. file systems, some NoSQL)</small>
- <small>**RDBMS** – manages data strictly in tables, with relationships enforced using keys (e.g. MySQL, PostgreSQL)</small>

<span style="font-size:1.15em">**5. What is MySQL?**</span>
MySQL is an open-source Relational Database Management System (RDBMS) that uses SQL to manage and query data.

**5.1. Key Features of MySQL**

- <small>**Open-source** – free to use and install</small>
- <small>**Easy to use** – simple to install and operate</small>
- <small>**Secure** – includes proper security for user data</small>
- <small>**Web-friendly** – works well with web stacks (e.g. LAMP: Linux, Apache, MySQL, PHP)</small>

**5.2. How to install/set up MySQL**

- <small>Download MySQL Community Server from the official site (dev.mysql.com)</small>
- <small>Run the installer, set a root password during setup</small>
- <small>Use **MySQL Workbench** (GUI) or the **command line** to connect and run queries</small>
- <small>Verify with: `mysql -u root -p`</small>
